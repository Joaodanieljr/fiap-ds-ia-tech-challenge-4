```python
import chromadb
from sentence_transformers import SentenceTransformer

modelo = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
colecao = chromadb.PersistentClient(path="../data/chroma").get_collection("avaliacoes_olist")

pergunta = "produto chegou quebrado ou com defeito"
vetor_pergunta = modelo.encode(pergunta, normalize_embeddings=True).tolist()

resultado = colecao.query(
    query_embeddings=[vetor_pergunta],
    n_results=5,
    where={"nota": {"$lte": 2}},   # só clientes insatisfeitos
)

for texto, meta, distancia in zip(resultado["documents"][0], resultado["metadatas"][0], resultado["distances"][0]):
    print(f"[{distancia:.3f}] nota {meta['nota']} | {meta['categoria']} | {texto[:100]}")
```

```python
pergunta = "qual é a capital da França?"
vetor_pergunta = modelo.encode(pergunta, normalize_embeddings=True).tolist()
resultado = colecao.query(query_embeddings=[vetor_pergunta], n_results=5)

for texto, distancia in zip(resultado["documents"][0], resultado["distances"][0]):
    print(f"[{distancia:.3f}] {texto[:100]}")
```
